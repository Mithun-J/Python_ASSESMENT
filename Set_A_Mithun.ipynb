```python
# Electricity Bill Calculator

customer_name = input()
units = int(input())

# Check for invalid units
if units < 0:
    print("Invalid units")

else:
    # Calculate electricity charge
    if units <= 100:
        electricity_charge = units * 2

    elif units <= 200:
        electricity_charge = (100*2) + ((units-100) * 3)

    elif units <= 400:
        electricity_charge = (100*2) + (100*3) + ((units-200) * 5)

    else:
        electricity_charge = (100*2) + (100*3) + (200*5) + ((units - 400) * 7)

    # Add service charge
    service_charge = 50
    total_bill = electricity_charge + service_charge

    # Display bill details
    print("Customer:", customer_name)
    print("Units Consumed:", units)
    print("Electricity Charge:", electricity_charge)
    print("Service Charge:", service_charge)
    print("Total Bill:", total_bill)
```

```python
# Student Result Management

N = int(input())

passed = 0
failed = 0
top_student = ""
highest_average = -1

results = []

for i in range(N):

    data = input().split()

    name = data[0]
    python_marks = int(data[1])
    database_marks = int(data[2])
    aptitude_marks = int(data[3])

    average = (python_marks + database_marks + aptitude_marks) / 3

    if (python_marks >= 40 and
        database_marks >= 40 and
        aptitude_marks >= 40 and
        average >= 50):

        passed = passed + 1

        if average >= 80:
            grade = "A"
        elif average >= 70:
            grade = "B"
        elif average >= 60:
            grade = "C"
        else:
            grade = "D"

        results.append(
            f"{name} - Average: {average:.2f} - Passed - Grade: {grade}"
        )

    else:
        failed = failed + 1

        results.append(
            f"{name} - Average: {average:.2f} - Failed"
        )

    if average > highest_average:
        highest_average = average
        top_student = name

# Print all results after taking all inputs
for result in results:
    print(result)

print()
print("Passed:", passed)
print("Failed:", failed)
print("Top Performer:", top_student)
```

```python
# Product class
class Product:

    def __init__(self, product_id, product_name, price, stock):
        self.product_id = product_id
        self.product_name = product_name
        self.price = price
        self.stock = stock


# Shopping cart class
class ShoppingCart:

    def __init__(self, products):
        self.products = products
        self.cart = {}

    # Add product to cart
    def add_product(self, product_id, quantity):

        if product_id not in self.products:
            print("Product not found")
            return

        if quantity <= 0:
            print("Invalid quantity")
            return

        product = self.products[product_id]

        if quantity > product.stock:
            print("Insufficient stock")
            return

        if product_id in self.cart:
            self.cart[product_id] += quantity
        else:
            self.cart[product_id] = quantity

        product.stock -= quantity

    # Remove product from cart
    def remove_product(self, product_id, quantity):

        if product_id not in self.cart:
            print("Product not in cart")
            return

        if quantity <= 0:
            print("Invalid quantity")
            return

        if quantity > self.cart[product_id]:
            print("Quantity exceeds cart quantity")
            return

        self.cart[product_id] -= quantity

        if self.cart[product_id] == 0:
            del self.cart[product_id]

        self.products[product_id].stock += quantity

    # Calculate total price
    def calculate_total(self):

        total = 0

        for product_id in self.cart:
            quantity = self.cart[product_id]
            product = self.products[product_id]

            total += product.price * quantity

        return total

    # Apply discount
    def apply_discount(self):

        total = self.calculate_total()

        if total >= 20000:
            discount_rate = 15
        elif total >= 10000:
            discount_rate = 10
        elif total >= 5000:
            discount_rate = 5
        else:
            discount_rate = 0

        discount_amount = total * discount_rate / 100
        final_amount = total - discount_amount

        return discount_rate, discount_amount, final_amount

    # Display cart details
    def display_cart(self):

        if len(self.cart) == 0:
            print("Cart is empty")
            return

        for product_id in self.cart:

            product = self.products[product_id]
            quantity = self.cart[product_id]

            subtotal = product.price * quantity

            print(
                f"{product.product_name} - "
                f"Quantity: {quantity} - "
                f"Subtotal: {subtotal:.0f}"
            )

        total = self.calculate_total()

        discount_rate, discount_amount, final_amount = self.apply_discount()

        print()
        print("Total Amount:", f"{total:.0f}")
        print("Discount:", str(discount_rate) + "%")
        print("Discount Amount:", f"{discount_amount:.0f}")
        print("Final Amount:", f"{final_amount:.0f}")


# Read products
n = int(input())

products = {}

for i in range(n):

    data = input().split()

    product_id = int(data[0])
    product_name = data[1]
    price = float(data[2])
    stock = int(data[3])

    product = Product(product_id, product_name, price, stock)

    products[product_id] = product


# Create shopping cart
cart = ShoppingCart(products)


# Read operations
operations = int(input())

for i in range(operations):

    data = input().split()

    operation = data[0]

    if operation == "ADD":

        product_id = int(data[1])
        quantity = int(data[2])

        cart.add_product(product_id, quantity)

    elif operation == "REMOVE":

        product_id = int(data[1])
        quantity = int(data[2])

        cart.remove_product(product_id, quantity)

    elif operation == "DISPLAY":

        cart.display_cart()

    elif operation == "TOTAL":

        print("Total:", f"{cart.calculate_total():.0f}")
```
